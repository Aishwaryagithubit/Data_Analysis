```python
Formula  = (weight in pounds*703)/ (height in inches * height in inches)
```

```python
name = input("please enter your name: ")
weight = float(input("please enter your weight: "))
feet =int(input("please enter your feet: "))
inches = int(input("please enter your inches: "))

height = (feet*12) + inches
BMI = (weight * 703)/(height * height)
```

```python
print(f"{name}, your BMI is: {BMI: .2f}")
```

```python
if BMI < 18.5:
    print(name  +", You are underweight.")
elif BMI < 24.9:
    print(name  +", You have a normal weight.")
elif BMI < 29.9:
    print(name  +", You are overweight.")
elif BMI < 34.9:
    print(name  +", You are obese.")
else:   
    print("Enter valid input")
```
