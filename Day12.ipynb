```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\ASUS\Desktop\DataPy\Superstore.csv", encoding = 'latin-1')
```

```python
# List of date columns
date_cols = ['Order Date', 'Ship Date']

# Loop through and convert each to datetime
for col in date_cols:
    df[col] = pd.to_datetime(df[col])

# Check dtypes
print(df[date_cols].dtypes)
```

```python
#unique values check
df["Row ID"].duplicated().sum()
```

```python
#Sales By Category
sales_cat = df.groupby("Category")["Sales"].sum()

ax = sales_cat.plot(kind="bar")

for container in ax.containers:
    ax.bar_label(container)
    
plt.title("Total Sales By Category")
plt.ylabel("Sales")
plt.show()
```

```python
profit_product = df.groupby("Sub-Category")["Profit"].sum().sort_values()

ax = profit_product.plot(kind="barh", figsize=(8,6))

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.title("Total Profit by Product")
plt.xlabel("Profit")
plt.ylabel("Product")
plt.show()
```
