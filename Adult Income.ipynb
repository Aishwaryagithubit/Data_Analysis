# Replace Salary Values ['<=50k', '>50k'] With O and 1

```python
df['income'].unique()
```

```python
df['income'].value_counts()
```

```python
sns.countplot(x='income', data=df)
plt.show()
```

```python
def income_data(sal):
    if sal == '50K':
        return 0
    else:
        return 1
```

```python
df['encoded_income'] = df['income'].apply(income_data)   #new column encode
```

```python
df.head(1)
```

```python
df.replace(to_replace=['<=50K', '>50K'], value=[0,1], inplace=True)
```
