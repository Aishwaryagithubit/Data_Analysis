```python
# List of date columns
date_cols = ['Order Date', 'Ship Date']

# Loop through and convert each to datetime
for col in date_cols:
    df[col] = pd.to_datetime(df[col])

# Check dtypes
print(df[date_cols].dtypes)
```

```python
#Number of Orders by Category
plt.figure(figsize = (4,4))
ax = sns.countplot(x = "Category", data = df)
ax.bar_label(ax.containers[0])
plt.title("Number of Orders by Category")
plt.show()
```

```python
#Sales By Category
sales_cat = df.groupby("Category")["Sales"].sum()

ax = sales_cat.plot(kind="bar")

for container in ax.containers:
    ax.bar_label(container)
    
plt.title("Total Sales By Category")
plt.ylabel("Sales")
plt.show()
```

```python
profit_product = df.groupby("Sub-Category")["Profit"].sum().sort_values()

ax = profit_product.plot(kind="barh", figsize=(8,6))

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.title("Total Profit by Product")
plt.xlabel("Profit")
plt.ylabel("Product")
plt.show()
```

```python
sales_segment = df.groupby(["Segment", "Category"])["Sales"].sum().reset_index()

ax = sns.barplot(data=sales_segment, x="Segment", y="Sales", hue="Category")

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.title("Customer Segment Sales by Category")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.legend(title="Category")
plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values()

ax = region_profit.plot(kind="bar")

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values()

ax = region_sales.plot(kind="bar")

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```
