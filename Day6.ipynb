```python
#MERGE , JOIN AND CONCATENATE
import pandas as pd
```

```python
df1 = pd.read_csv(r"C:\Users\ASUS\Desktop\DataPy\patients.csv")
```

```python
df1
```

```python
df2 = pd.read_csv(r"C:\Users\ASUS\Desktop\DataPy\appointments.csv")
df2
```

```python
df1.merge(df2)  #inner join
```

```python
df1.merge(df2, how = 'inner')
```

```python
df1.merge(df2, how = 'inner', on ='patient_id')
```

```python
df1.merge(df2, how = 'outer')
```

```python
merged_df = df1.merge(df2, how='left', on='patient_id')
merged_df
```

```python
merged_df = df1.merge(df2, how='right', on='patient_id')
merged_df
```

```python
df = df1.merge(df2, how='cross')
df
```

```python
df1.join(df2, on = 'patient_id', how = 'outer', lsuffix = '_Left',rsuffix = '_Right') 
```

```python
df4 = df1.set_index('patient_id').join(df2.set_index('patient_id'), lsuffix = '_Left', rsuffix = '_Right', how = 'outer')
df4
```

```python
pd.concat([df1,df2], join = 'inner')
```

```python
pd.concat([df1,df2], join = 'outer')
```

```python
pd.concat([df1,df2], join = 'inner', axis = 1)
```

```python
pd.concat([df1,df2])
```
