# Group By

```python
region_sales = df.groupby("region") ["sales_amount"].mean()   #single column grouping
print(region_sales)
```

```python
cat_region = df.groupby(["category","region"]) ["sales_amount"].sum()  # multi column grouping ("Hierarchical Index) 
print(cat_region)
```

```python
pivoted_view = cat_region.unstack(level="region")  #Unstacking into tabular format
print(pivoted_view)
```

# Pivot_table & crosstab()

```python
pivot_df = pd.pivot_table(
    df,
    values="sales_amount",
    index="customer_segment",
    columns="category",
    aggfunc="mean",
    fill_value=0.0,
    margins=True
)

print(pivot_df)                                
```

```python
freq_table = pd.crosstab(index=df["region"],columns=df["customer_segment"],normalize="index")
print(freq_table)
```

```python
print(freq_table.round(3)*100)
```

# Unpivot Data With melt()

```python
wide_df = pd.DataFrame({
    "Product":["A","B"],
    "2024":[100,300],
    "2025":[150,350]})

long_df = pd.melt(
    wide_df, id_vars =["Product"],
    value_vars = ["2024", "2025"],
    var_name="Year",
    value_name="Sales")

print(long_df)
```

# Statistical Insight Extraction & Outliers

# Missing Data Remediation : Deletion vs Imputation

```python
#Strategy A : Row Deletion (While missing records)
df_clean = df.dropna(subset =["user_id"])

#Strategy B : Mean & Mode Imputation
spend_median = df["monthly_spend"].median()
df["monthly_spend"] = df["monthly_spend"].fillna(spend_median)

cat_mode = df["category"].mode()[0]
df["category"] = df["category"].fillna(cat_mode)

#Strategy C: Forward Fill for Time Series/sensor Stream
df["sensor_reading"] = df["sensor_reading"].ffill().bfill()
```
