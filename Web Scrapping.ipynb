```python
#BeautifulSoup and Requests

import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://www.reddit.com/'
```

```python
page = requests.get(url)
```

```python
soup = BeautifulSoup(page.text, 'html.parser')   # use .text and specify parser
print(soup.prettify())
```

```python
#Find and Find_all

import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://www.reddit.com/'
```

```python
page = requests.get(url)
```

```python
soup = BeautifulSoup(page.text, 'html')
print(soup)
```

```python
soup.find('div')
```

```python
soup.find_all('div')
```

```python
soup.find_all('div', class_ = 'col-md-12')
```

```python
soup.find_all('p')
```

```python
soup.find_all('p', class_ = 'logo')
```

```python
soup.find('p', class_ = 'lead').text.strip()
```

```python
soup.find_all('td')
```
