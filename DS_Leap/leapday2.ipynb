# 2D Slicing

```python
matrix = np.random.rand(10,10)
print(matrix)
```

```python
row_zero = matrix[0:1]
print(row_zero)
```

```python
col_five = matrix[:,5]
print(col_five)
```

```python
top_corner= matrix[0:3]
print(top_corner)
```

```python
bottom_corner = matrix[-3:, -3:]
print(bottom_corner)
```

```python
bottom_corner = matrix[7:,7:]
print(bottom_corner)
```

```python
#first element
a = matrix[0:1,0:1]
print(a)
```

```python
print(a.ndim)
```

# Value Retriving
