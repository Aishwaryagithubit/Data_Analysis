# Task 1.1: Environmental Sensor Analytics

```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
raw_temperatures = np.random.normal(loc=24.0, scale=3.0, size=72)
```

```python
# Simulated corrupt readings
raw_temperatures[[5, 12, 28, 45, 60]] = [99.9, -15.0, 88.8, -50.0, 102.4]
print(raw_temperatures)
```

```python
#create boolean mask
mask = (raw_temperatures >= 15.0) & (raw_temperatures <= 40.0)
print(mask)
```

```python
# Replace invalid values with median of valid temperature
median_temp = np.median(raw_temperatures[mask])
raw_temperatures[~mask] = median_temp
print(raw_temperatures)
```

```python
# Statistics
mean = np.mean(raw_temperatures)
std = np.std(raw_temperatures)
variance = np.var(raw_temperatures)
```

```python
# Print results
print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))
print("Variance:", round(variance, 2))
```

# Part 2: Sliding Median Filter

```python
#create original noisy signal
signal = np.array([10, 11, 99, 12, 13, -45, 14, 15, 16])
print(f"Original Signal: {signal}")
```

```python
#make a copy to store filter values
filtered_signal = signal.copy()
print(f"Filtered Signal:{filtered_signal}")
```

```python
# Apply a sliding median filter using a window of size 3
for i in range(1,len(signal)-1):
    filtered_signal[i] = np.median(signal[i-1:i+2])

print("Original Signal:", signal)
print("Filtered Signal:", filtered_signal)
```
