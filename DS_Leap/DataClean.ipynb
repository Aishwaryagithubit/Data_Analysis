# Boolean Masking

```python
data = np.random.randn(100)
print(data)
```

```python
data = np.random.randn(100)
mask = data > 0
pos_data = data[mask]
```

```python
print(pos_data)
```

```python
mask2 = data < 0
neg_data = data[mask2]
print(neg_data)
```

```python
#remove outlier
print(data [data > 3])
```

```python
print(data[data < -3])
```

# Matrix Algebra

```python
# np.linalg for heavy computation
a = np.array([[1,2],[3,4]])
print(a)
```

```python
b = np.linalg.inv(a)   #matrix Inverse
print(b)
```

```python
I = a @ b   #matrix Multiplication (identity matrx)
print(I)
```

```python
eig_vals = np.linalg.eigvals(a)
print(eig_vals)
```

```python
j = a * b
print(j) 
```
