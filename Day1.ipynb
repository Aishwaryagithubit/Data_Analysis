```python
#combined unique log
daily_log = {1,3,5,7,3,24,5,7,3,2,0}
wifes_log = {1,2,31,4,1,2,5,6,3,2}
print(daily_log | wifes_log)
```

```python
#only selects match
print(daily_log & wifes_log)
```

```python
#remove value others keeping other same
print(daily_log - wifes_log)
```

```python
#value in one or other but not in both
print(daily_log ^ wifes_log)
```

```python
number = 0
while number < 5:
    print(number)
    if number == 6:
        break
    number = number + 1
else:
    print("no longer < 5")
```

```python
number = 0

while number < 5:
    number = number + 1
    if number == 3:
         continue
         print(number)
else:
         print("no longer < 5") #infinite becoz stop that condition else all run
```
